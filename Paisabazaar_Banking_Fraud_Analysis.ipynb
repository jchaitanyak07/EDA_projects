## 4. Data Cleaning

### Removing unneccessary columns --> ID, Customer_ID, Name, SSN

```python
data = data.drop(columns = ["ID", "Customer_ID", "Name", "SSN"])
```

```python
data.columns
```

### Converting 'Age' Column from float to integer

```python
data['Age'].head()
```

```python
data['Age'] = data['Age'].astype(int)
```

```python
data["Age"].head()
```

```python
data['Age'].info()
```

### Converting other columns from float to integer

```python
float_col_list = ['Num_Bank_Accounts', "Num_Credit_Card", "Num_of_Loan", "Delay_from_due_date", "Num_of_Delayed_Payment", "Num_Credit_Inquiries", "Credit_History_Age"]

for col in float_col_list:
  data[col] = data[col].astype(int)
```

```python
data["Num_Bank_Accounts"].info()
```

### Stripping Object Columns

```python
strip_col_list = ["Occupation", "Type_of_Loan", "Credit_Mix", "Payment_of_Min_Amount", "Payment_Behaviour", "Credit_Score"]
```

```python
for col in strip_col_list:
  data[col] = data[col].apply(lambda string : string.strip())
```

### Number of bank accounts = 0

Dropping such rows where number of bank accounts is itself is 0.

```python
data[data["Num_Bank_Accounts"] == 0]
```

```python
data = data[data["Num_Bank_Accounts"] > 0]
```

```python
data.shape
```

### Payment_of_Min_Amount --> has 'NM' values

```python
data["Payment_of_Min_Amount"].unique()
```

```python
mode = data["Payment_of_Min_Amount"].mode()[0]

mode
```

```python
data["Payment_of_Min_Amount"] = data["Payment_of_Min_Amount"].replace('NM', mode)
```

```python
data["Payment_of_Min_Amount"].unique()
```

### Num_Credit_Card = 0

```python
data = data[data["Num_Credit_Card"] > 0]
```

```python
data.shape
```

## 5. Adding new columns for better analysis

### 'Customer_risk' columns

```python
def risk_func(score):
  if score == "Standard":
    return "Medium Risk"
  elif score == "Poor":
    return "High Risk"
  else:
    return "Low Risk"
```

```python
data["Customer_risk"] = data["Credit_Score"].apply(risk_func)

data["Customer_risk"].head()
```

```python
data["Customer_risk"].value_counts()
```

```python
data.head()
```

### Debt_to_income_ratio --> Outstanding debt divided by Annual Income


Debt-to-Income Ratio (DTI) --> amount of debt an individual or organization has relative to their income

```python
data["Debt_to_income_ratio"] = round((data["Outstanding_Debt"] / data["Annual_Income"]) * 100, 4)

data["Debt_to_income_ratio"].head()
```

```python
data["Debt_to_income_ratio"].describe()
```

### EMI_burden --> what % of monthly salary goes to EMI

```python
data["EMI_burden_pct"] = round((data["Total_EMI_per_month"] / data["Monthly_Inhand_Salary"]) * 100, 2)

data["EMI_burden_pct"].head()
```

```python
data["EMI_burden_pct"].describe()
```

### Age_group

```python
def get_age_group(age):
  if age < 25:
    return "Under 25"
  elif age < 35:
    return "25-34"
  elif age < 45:
    return "35-44"
  else:
    return "45+"
```

```python
data["Age_group"] = data["Age"].apply(get_age_group)

data["Age_group"].head()
```

```python
data["Age_group"].value_counts()
```
